> **A. Calcular la probabilidad de que al tomar un conjunto de 3 alumnos de un grupo de 40, los 3 sean mujeres, si sólo hay 9 mujeres en el grupo.**

```python
#Usaremos las convinatorias, para ello es importante definir el el coeficiente binomial
```

$$
\binom{n}{k} = \frac{n!}{k!(n-k)!}
$$

Las diferentes formas de escoger un grupo de 3 de un grupo de 40 alumnos
$$
\binom{40}{3} = \frac{40!}{3!(40-3)!}
$$

```python
convinatorias_grupal = mt.factorial(40)/(mt.factorial(3)*(mt.factorial(37)))
convinatorias_grupal
```

Ahora los casos a favor
$$
\binom{9}{3} = \frac{9!}{3!(9-3)!}
$$

```python
convinatorias_favor = mt.factorial(9)/(mt.factorial(3)*(mt.factorial(6)))
convinatorias_favor
```

Ahora simplemente podemos hacer el calculo  de probabilidad
$$
P = \frac{C.F}{C.T}
$$

```python
P = convinatorias_favor/convinatorias_grupal
print(f"La probabilidad de escoger a 3 mujeres es de  {P}")
```

> **B. Considere una clase de 100 alumnos. Asigne, de manera aleatoria, sus calificaciones entre 5 y 10. Suponga que su resultado sea típico. a) Calcule el promedio y la varianza del experimento**

```python
#Primero definimos nuestros numeros aleatorios en un rango del 5 al 10
calificaciones = np.random.randint(5,11 , size=100)
calificaciones
```

```python
print(f"El promedio es de {np.mean(calificaciones)}")
print(f"La varianza es de {np.var(calificaciones)}")
```

> **b. repita el ejercicio 10 veces (10 cursos)**

```python
def cal(min,max,n):
  calificaciones = np.random.randint(min,max+1,n)
  promedio = np.mean(calificaciones)
  varianza = np.var(calificaciones)
  return promedio,varianza
```

```python
promedios = []
varianzas = []
for i in range (10):
  promedios.append(cal(5,10,100)[0])
  varianzas.append(cal(5,10,100)[1])
  print(f"Para el evento {i+1} el promedio es {promedios[i]} y la varianza {varianzas[i]}")
print(promedios )



```

>**C. De A2, calcular promedio y desviación estándar.**

```python
#Calculamos el promedio de los promedios
print(f"EL promedio de los promedios es{np.mean(promedios)}")
#La desviacion estandar es
print(f"La desviacion estandar de los promedios es {np.std(promedios)}")
```
