> Use Monte Carlo integration to find the volume of a hemisphere
of radius 1.

We can use the symmetry of the problem to find the volume
of one quarter (the positive {x, y, z} quarter) of the hemisphere,

z range from 0–1. This range of values corresponds exactly to the
evenly-distributed random numbers given by random.random().

## **Problem 6-0**

> Calculate $$ \int_a^b sin(x) \, dx$$ using Monte Carlo techniques. Report your
uncertainty in the result, and compare with the known result.

Usaremos
$$
\int_a^b sin(x) \, dx = \frac{(b-a)}{N} \sum_{n=o}^N f(x_i)
$$



```python
import numpy as np
#Definimos la funcion f(x)
def f(x):
  return np.sin(x)
a,b = 0 , np.pi #Definimos los limites de intregracion

N = 1000000 #Cantidad de puntos randoms
puntos_randoms = np.random.uniform(a,b,N)
#Evaluamos la funcion
funcion = f(puntos_randoms)
#Valor de la integral por M.C
integral = (b-a)/N * np.sum(funcion)
#Calculamos la incertidumbre
incertidumbre = (b - a) * np.std(funcion) / np.sqrt(N)
valor_teorico = -np.cos(b) + np.cos(a)

print(f"Valor de M.C con incertidumbre {integral} +- {incertidumbre}")
print(f"Valor Teorico {valor_teorico}")
```
